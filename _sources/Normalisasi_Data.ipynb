## 1. Pengertian Normalisasi Data

Normalisasi data adalah proses mengubah atau menyesuaikan nilai data agar memiliki skala yang lebih seragam, sehingga data lebih mudah diolah oleh algoritma data mining atau machine learning.

Biasanya data yang dikumpulkan masih berupa data mentah yang dibuat untuk kebutuhan sistem atau berasal dari sumber tertentu. Nilai data tersebut sering memiliki rentang yang berbeda-beda, sehingga dapat mempengaruhi hasil analisis atau model prediksi.

Oleh karena itu, dilakukan normalisasi untuk mengubah nilai data ke dalam skala tertentu tanpa mengubah informasi penting dari data tersebut. Proses ini tidak menambah atribut baru, tetapi hanya mengubah nilai dari atribut yang sudah ada agar memiliki distribusi atau skala yang lebih sesuai.

## 2. Macam-Macam Normalisasi Data

### 1. Min-Max Normalization
Min-Max Normalization adalah metode normalisasi yang digunakan untuk mengubah nilai data ke dalam rentang tertentu, biasanya 0 sampai 1.
Metode ini dilakukan dengan cara menyesuaikan nilai data berdasarkan nilai minimum dan maksimum dari suatu atribut.

Tujuannya adalah agar semua nilai data memiliki skala yang sama, sehingga tidak ada data yang terlalu besar atau terlalu kecil ketika diproses oleh algoritma.

Rumus Min-Max Normalization:
$$
v' = \frac{v - \min(A)}{\max(A) - \min(A)} \times (new\_maxA - new\_minA) + new\_minA
$$

Keterangan:

* v
v = nilai data asli
* min⁡(A)
min(A) = nilai minimum atribut A
* max⁡(A)
max(A) = nilai maksimum atribut A
* v′
v
′
 = nilai setelah normalisasi
* new_minA
new_minA = batas minimum baru
* new_maxA
new_maxA = batas maksimum baru

#### Contoh Min-Max Normalization

Data awal:
| Nilai |
| ----- |
| 10    |
| 20    |
| 30    |

Diketahui:
* nilai minimum = 10
* nilai maksimum = 30
* rentang normalisasi = 0 sampai 1

Hasil normalisasi:
| Nilai Asli | Nilai Normalisasi |
| ---------- | ----------------- |
| 10         | 0                 |
| 20         | 0.5               |
| 30         | 1                 |

### 2. Z-Score Normalization
Sebagai alternatif, digunakan Z-Score Normalization. Metode ini menormalisasi data dengan menggunakan rata-rata (mean) dan standar deviasi dari data.

Rumus Z-Score Normalization:
$$
v' = \frac{v - \bar{A}}{\sigma_A}
$$

Keterangan:
* vv = nilai data asli
* Aˉ= rata-rata (mean) dari atribut A
* σA = standar deviasi atribut A
* v′ = nilai setelah normalisasi

Setelah dilakukan normalisasi dengan metode ini, data akan memiliki:
* rata-rata = 0
* standar deviasi = 1

#### Rumus Mean
$$
\bar{A} = \frac{1}{n}\sum_{i=1}^{n} v_i
$$
#### Rumus Standar Deviasi
$$
\sigma_A = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(v_i - \bar{A})^2}
$$
#### Contoh Z-score
Data awal:
| Data |
| ---- |
| 10   |
| 20   |
| 30   |

Hasil Normalisasi:
| Nilai Asli | Z-Score |
| ---------- | ------- |
| 10         | -1.22   |
| 20         | 0       |
| 30         | 1.22    |

## 3. Implementasi Normalisasi Menggunakan Python

### 1. Min-Max Normalization Menggunakan Library sklearn
Contoh Min-Max Normalization menggunakan library sklearn:

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

data = pd.DataFrame({
    'umur':[10,20,30],
    'gaji':[5,8,10]
})

scaler = MinMaxScaler()

normalized = scaler.fit_transform(data)

print(normalized)
```

### 2. Z-Score Normalization Menggunakan Library sklearn
Contoh Z-Score Normalization menggunakan library sklearn:

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

data = pd.DataFrame({
    'umur':[10,20,30],
    'gaji':[5,8,10]
})

normalized = scaler.fit_transform(data)
print(normalized)
```

### 3. Decimal Scaling Normalization Menggunakan Function
Contoh Decimal Scaling Normalization menggunakan Function:

```python
import pandas as pd
import numpy as np

data = pd.Series([210,350,600])

def decimal_scaling(x):
    j = len(str(int(x.abs().max())))
    return x / (10**j)

hasil = decimal_scaling(data)

print(hasil)
```
