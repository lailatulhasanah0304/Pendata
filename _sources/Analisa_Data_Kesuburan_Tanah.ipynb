# **Analisa Data Kesuburan Tanah**

## 1. Pendahuluan

Analisis data kesuburan tanah dilakukan untuk mengklasifikasikan kondisi tanah menjadi dua kelas, yaitu Subur dan Tidak Subur. Metode yang digunakan adalah K-Nearest Neighbors (KNN) dengan bantuan software KNIME Analytics Platform. KNN dipilih karena mampu melakukan klasifikasi berdasarkan kedekatan jarak antar data.

image.png

## 2. Dataset

Dataset terdiri dari 2000 data dengan 10 fitur dan 1 label kelas.

### Fitur:
* pH Tanah
* N Total (%)
* P Tersedia (ppm)
* K Tersedia (meq/100g)
* C Organik (%)
* KTK (meq/100g)
* Kejenuhan Basa (%)
* Tekstur Tanah (kategori)
* Kadar Air (%)
* Bulk Density (g/cm³)

### Label:
* Subur
* Tidak Subur

## 3. Tahapan Pemrosesan Data di KNIME
### 3.1 Missing Value Handling

Data kosong ditangani dengan:
* Numerik → Mean
* Kategorikal → Mode

Tujuan:

Menghindari bias dan error pada perhitungan jarak

### 3.2 One to Many

Fitur Tekstur Tanah merupakan data kategorikal sehingga perlu diubah menjadi bentuk numerik menggunakan teknik One-Hot Encoding (One to Many). Hal ini dilakukan karena algoritma KNN hanya dapat memproses data numerik.

### 3.3 Normalisasi Data

Karena KNN berbasis jarak, maka data dinormalisasi.

Rumus Min-Max Normalization:

image.png

Tujuan:
* Menyamakan skala fitur
* Menghindari dominasi fitur tertentu

## 4. Penerapan K-Nearest Neighbors (KNN)

Algoritma KNN digunakan untuk mengklasifikasikan data berdasarkan kedekatan dengan data tetangga terdekat.

Parameter yang digunakan:

K = 5

Distance Metric = Euclidean Distance

Model dilatih menggunakan data training dan kemudian digunakan untuk memprediksi data testing.


## 5. Hasil Evaluasi Model

Berdasarkan hasil dari Confusion Matrix, diperoleh data sebagai berikut:
| Aktual / Prediksi | Tidak Subur | Subur |
| ----------------- | ----------- | ----- |
| Tidak Subur       | 105         | 82    |
| Subur             | 70          | 86    |


image.png


## 6. Perhitungan Metrik Evaluasi
### **Accuracy**
Accuracy = (TP + TN) / (TP + TN + FP + FN)

Accuracy = (86 + 105) / (86 + 105 + 82 + 70)

Accuracy = 191 / 343

Accuracy = 0.557 ≈ 55.7%


### **Precision**
Precision = TP / (TP + FP)

Precision = 86 / (86 + 82)

Precision = 86 / 168

Precision = 0.512 ≈ 51.2%

### **Recall**
Recall = TP / (TP + FN)

Recall = 86 / (86 + 70)

Recall = 86 / 156

Recall = 0.551 ≈ 55.1%

### **F1-Score**
F1-Score = 2 × (Precision × Recall) / (Precision + Recall)

F1-Score = 2 × (0.512 × 0.551) / (0.512 + 0.551)

F1-Score = 0.531 ≈ 53.1%

## 7. Analisis Hasil
* Model memiliki akurasi sebesar 55.7%, yang menunjukkan performa masih tergolong rendah.
* Nilai precision (51.2%) menunjukkan bahwa prediksi “Subur” masih cukup banyak yang salah.
* Nilai recall (55.1%) menunjukkan model belum optimal dalam mendeteksi seluruh data tanah yang benar-benar subur.
* Hal ini bisa disebabkan oleh:
  * Parameter K yang belum optimal
  * Data yang masih kompleks
  * Distribusi fitur yang belum optimal


## 8. Kesimpulan

Berdasarkan hasil analisis menggunakan metode KNN, model mampu melakukan klasifikasi kesuburan tanah namun dengan tingkat akurasi yang masih rendah. Oleh karena itu, diperlukan optimasi lebih lanjut seperti penentuan nilai K yang optimal atau penggunaan metode lain seperti Random Forest atau Decision Tree untuk meningkatkan performa model.
